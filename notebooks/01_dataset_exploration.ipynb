## 2. Inspect Target Label Distribution & Class Imbalance

```python
label_col = "Label"
counts = df[label_col].value_counts()
proportions = df[label_col].value_counts(normalize=True) * 100

summary_df = pd.DataFrame({"Count": counts, "Percentage (%)": proportions.round(2)})
summary_df
```
